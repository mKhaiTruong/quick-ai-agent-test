## Intro

```python
from pydantic import BaseModel
from typing import List, Dict, Optional

class PatientData(BaseModel):
    name:       str
    age:        int
    weight:     float
    married:    bool = False
    allergies:  Optional[List[str]] = None
    contact_info: Dict[str, str]
```

```python
def add_patient_data(patient: PatientData):
    print("Name = ", patient.name)
    print("Age = ", patient.age)
    print("Weight = ", patient.weight)
    print("Married = ", patient.married)
    print(patient.allergies)
    print("Data added successfully to the database!")

patient_data = {
    "name": "MKHUE", 
    "age": "23", 
    "weight": "73.5", 
    "contact_info": {"email": "abc@gmail.com", "phone": "123-456-7890"}
}
p1 = PatientData(**patient_data)
add_patient_data(p1)
```

## Field Validator

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field, field_validator
from typing import List, Dict, Optional, Annotated

class PatientData(BaseModel):
    name:           str = Field(max_length=50)
    email:          EmailStr
    age:            int = Field(gt=0, lt=100)
    weight:         float
    married:        bool = False
    allergies:      Optional[List[str]] = Field(max_length=5, default=None)
    contact_info:   Dict[str, str]
    
    @field_validator("email")
    @classmethod
    def email_validator(cls, value):
        valid_domains = ['hdfc.com', 'icici.com']
        domain_name   = value.split('@')[1]
        
        if domain_name not in valid_domains:
            raise ValueError(f"Invalid email domain: {domain_name}")
        
        return value
    
    @field_validator("name")
    @classmethod
    def transform_name(cls, value):
        return value.upper()
```

```python
patient_data = {
    "name": "MKHUE", 
    "email": "minhkhai0402@hdfc.com", 
    "linkedin_url": "https://www.linkedin.com/in/tdmk/", 
    "age": "23", 
    "weight": 73.5, 
    "married": "False",
    "allergies": ["toxic", "snakes"],
    "contact_info": {"phone": "123-456-7890"}}

p1 = PatientData(**patient_data)
add_patient_data(p1)
```

## Computed Fields

```python
from pydantic import BaseModel, EmailStr, Field, computed_field
from typing import List, Dict, Optional, Annotated

class PatientData(BaseModel):
    name:           str = Field(max_length=50)
    email:          EmailStr
    age:            int = Field(gt=0, lt=100)
    weight:         float
    height:         float
    married:        bool = False
    allergies:      Optional[List[str]] = Field(max_length=5, default=None)
    contact_info:   Dict[str, str]
    
    @computed_field
    @property
    def bmi(self) -> float:
        bmi = round(self.weight/(self.height**2),2)
        return bmi
```

```python
patient_data = {
    "name": "MKHUE", 
    "email": "minhkhai0402@hdfc.com", 
    "linkedin_url": "https://www.linkedin.com/in/tdmk/", 
    "age": "23", 
    "weight": 73.5,
    "height": 1.8,
    "married": "False",
    "allergies": ["toxic", "snakes"],
    "contact_info": {"phone": "123-456-7890"}}

p1 = PatientData(**patient_data)
add_patient_data(p1)
print("BMI = ", p1.bmi)
```
